```python
from sklearn.preprocessing import StandardScaler
scale=StandardScaler()
df['Amount_scale']=scale.fit_transform(df[['Amount']])
```

```python
df.head()
```

```python
df=df.drop(['Time'],axis=1)
```

```python
df.head()
```

```python
# independent and dependent featyres
X=df.drop('Class',axis=1)
Y=df['Class']
```

```python
sample = df.sample(1).iloc[0]
sample
```

```python
# train test split
from sklearn.model_selection import train_test_split
X_train,X_test,Y_train,Y_test=train_test_split(X,Y,test_size=0.20,random_state=42)
```

```python
# using smote balanced training dataset
import imblearn
from imblearn.over_sampling import SMOTE
oversample=SMOTE()
X_train_res,Y_train_res=oversample.fit_resample(X_train,Y_train)
```

```python
X_train.shape
```

```python
X_train_res.shape
```

```python
Y_train.shape
```

```python
Y_train_res.shape
```

```python
plt.figure(figsize=(15,9))
plt.subplot(1,2,1)
sns.countplot(x=Y_train,palette='coolwarm')
plt.title("before smote (imbalnced)")
plt.subplot(1,2,2)
sns.countplot(x=Y_train_res,palette='viridis')
plt.title("after smote (balnced)")
```

```python
# train model
from sklearn.linear_model import LogisticRegression
logistic=LogisticRegression()
logistic.fit(X_train_res,Y_train_res)
```

```python
y_pred=logistic.predict(X_test)
```

```python
y_pred
```

```python
plt.scatter(Y_test,y_pred)
```

```python
from sklearn.metrics import accuracy_score,confusion_matrix,classification_report
score=accuracy_score(Y_test,y_pred)
cm=confusion_matrix(Y_test,y_pred)
cr=classification_report(Y_test,y_pred)
```

```python
print(score)
print(cm)
print(cr)
```

```python
sns.heatmap(cm,annot=True)
plt.xlabel("Predicted")
plt.ylabel("Actual")
plt.title("Confusion Matrix")
plt.show()
```

```python
# threshold tunning
Y_prob=logistic.predict_proba(X_test)[:,1]
```

```python
threshold=0.8
y_pred_new=(Y_prob>threshold).astype(int)
```

```python
print(confusion_matrix(Y_test,y_pred_new))
print(classification_report(Y_test,y_pred_new))
```

```python
from sklearn.metrics import roc_auc_score
roc_auc_score(Y_test, Y_prob)
```

```python
from sklearn.metrics import roc_curve
fpr, tpr, thresholds = roc_curve(Y_test, Y_prob)
plt.plot(fpr, tpr, label="ROC Curve")
plt.plot([0,1], [0,1], linestyle='--')
plt.xlabel("False Positive Rate")
plt.ylabel("True Positive Rate (Recall)")
plt.title("ROC Curve")

plt.legend()
plt.show()
```

```python
from sklearn.metrics import precision_recall_curve
precision, recall, thresholds = precision_recall_curve(Y_test, Y_prob)
plt.plot(recall, precision)
plt.xlabel("Recall")
plt.ylabel("Precision")
plt.title("Precision-Recall Curve")
plt.show()
```

```python
from sklearn.ensemble import RandomForestClassifier

rf = RandomForestClassifier(
    n_estimators=100,
    max_depth=10,
    class_weight={0:1, 1:3},
    random_state=42
)

rf.fit(X_train_res, Y_train_res)

y_prob_rf = rf.predict_proba(X_test)[:,1]

threshold = 0.8
y_pred_rf = (y_prob_rf > threshold).astype(int)

from sklearn.metrics import classification_report
print(classification_report(Y_test, y_pred_rf))
```

```python
from sklearn.tree import DecisionTreeClassifier

dt = DecisionTreeClassifier(
    max_depth=5,                 
    class_weight={0:1, 1:3},     
    random_state=42,
)

dt.fit(X_train_res, Y_train_res)
y_prob_dt = dt.predict_proba(X_test)[:, 1]
threshold = 0.8
y_pred_dt = (y_prob_dt > threshold).astype(int)
from sklearn.metrics import classification_report
print(classification_report(Y_test, y_pred_rf))
```

```python
from xgboost import XGBClassifier

xgb = XGBClassifier(
    n_estimators=200,
    max_depth=6,
    learning_rate=0.1,
)

xgb.fit(X_train_res, Y_train_res)

y_prob_xgb = xgb.predict_proba(X_test)[:,1]
y_pred_xgb = (y_prob_xgb > 0.8).astype(int)
```

```python
from sklearn.metrics import classification_report, roc_auc_score, confusion_matrix

print(confusion_matrix(Y_test, y_pred_xgb))
print(classification_report(Y_test, y_pred_xgb))
print("ROC AUC:", roc_auc_score(Y_test, y_prob_xgb))
```
